```python
from faker import Faker
import random
import pandas as pd

# dados em português
fake = Faker('pt_br')
```

```python
DADOS_GEOGRAFICOS = {
    "SP": ["São Paulo", "Campinas", "Santos", "São Bernardo do Campo", "Sorocaba"],
    "RJ": ["Rio de Janeiro", "Niterói", "Duque de Caxias", "Petrópolis", "Búzios"],
    "MG": ["Belo Horizonte", "Uberlândia", "Ouro Preto", "Juiz de Fora", "Contagem"],
    "RS": ["Porto Alegre", "Gramado", "Caxias do Sul", "Canoas", "Pelotas"],
    "PR": ["Curitiba", "Londrina", "Maringá", "Foz do Iguaçu", "Cascavel"],
    "BA": ["Salvador", "Porto Seguro", "Feira de Santana", "Ilhéus", "Vitória da Conquista"],
    "PE": ["Recife", "Olinda", "Caruaru", "Petrolina", "Jaboatão dos Guararapes"]
}

def cliente():
    # Definição de Gênero e Nome
    genero = random.choice(["M", "F"])
    if genero == "M":
        nome = fake.name_male()
        genero = "Masculino"
    else:
        nome = fake.name_female()
        genero = "Feminino"

    # Sorteio Consistente de Estado e Cidade
    # Sorteia uma sigla de estado do nosso dicionário
    estado = random.choice(list(DADOS_GEOGRAFICOS.keys()))
    # Sorteia uma cidade que pertence APENAS a esse estado sorteado
    cidade = random.choice(DADOS_GEOGRAFICOS[estado])

    # Dicionário com as informações
    usuario = {
        "nome": nome,
        "email": fake.email(),
        "cidade": cidade,
        "estado": estado,
        "genero": genero,
        # Data de nascimento de 18 a 75 anos
        "data_nascimento": fake.date_of_birth(
            minimum_age=18, maximum_age=75
        ).strftime("%d/%m/%Y"),
        # Cadastro nos últimos 3 anos
        "data_cadastro": fake.date_between(
            start_date="-3y", end_date="today"
        ).strftime("%d/%m/%Y"),
    }

    return usuario
```

```python
# Gerando 100 clientes e transformando em um DF
lista_usuarios = [cliente() for _ in range(100)]
df_clientes = pd.DataFrame(lista_usuarios)
```

```python
# Criando uma coluna de ids
df_clientes["id"] = range(1, len(df_clientes) + 1)
```

```python
# reorganizando e exportando
df_clientes = df_clientes[['id', 'nome', 'email', 'cidade', 'estado', 'genero', 'data_nascimento', 'data_cadastro']]
df_clientes.to_excel('clientes.xlsx', index=False)
```

# Vendedores

```python
REGIOES = ["Norte", "Nordeste", "Centro-Oeste", "Sudeste", "Sul"]
METAS = [15000, 20000, 25000, 30000, 40000, 50000]

def vendedor():
    nome = fake.name()

    # Cria um e-mail corporativo baseado no nome do vendedor
    nome_limpo = (
        nome.lower()
        .replace(" ", ".")
        .encode("ascii", "ignore")
        .decode("utf-8")
    )
    email_corporativo = f"{nome_limpo}@empresa.com.br"

    regiao = random.choice(REGIOES)
    meta = random.choice(METAS)


    vendedor = {
        "nome": nome,
        "email_corporativo": email_corporativo,
        "regiao": regiao,
        "meta_vendas": meta,
    }
    return vendedor
```

```python
# Gerando 15 vendedores e transformando em um DF
lista_vendedor = [vendedor() for _ in range(15)]
df_vendedores = pd.DataFrame(lista_vendedor)
```

```python
# Criando uma coluna de ids
df_vendedores["id"] = range(1, len(df_vendedores) + 1)
```

```python
# Reorganizando e exportando
df_vendedores = df_vendedores[['id', 'nome', 'email_corporativo', 'regiao', 'meta_vendas']]
df_vendedores.to_excel('vendedores.xlsx', index=False)
```

# Pedidos

```python
STATUS_OPCOES = ["Pendente", "Enviado", "Entregue", "Cancelado"]

# Pesos para o status
PESOS_STATUS = [0.15, 0.20, 0.55, 0.10] 

def pedido():
    # Sorteia um cliente entre os 100 existentes
    cliente_id = random.randint(1, 100)
    
    # Sorteia um vendedor entre os 20 existentes
    vendedor_id = random.randint(1, 20)
    
    # Gera uma data aleatória nos últimos 3 anos
    data_pedido = fake.date_between(start_date="-3y", end_date="today").strftime("%d/%m/%Y")
    
    # Sorteia o status usando os pesos definidos acima
    status = random.choices(STATUS_OPCOES, weights=PESOS_STATUS, k=1)[0]

    pedido = {
        "cliente_id": cliente_id,
        "vendedor_id": vendedor_id,
        "data": data_pedido,
        "status": status
    }
    
    return pedido
```

```python
# Gera uma lista com 1000 pedidos e transformando em df
lista_pedidos = [pedido() for _ in range(1000)]
df_pedidos = pd.DataFrame(lista_pedidos)
```

```python
# Adicionando a coluna de id
df_pedidos.insert(0, "id", range(1, len(df_pedidos) + 1))
```

```python
# Exportando os pedidos
df_pedidos.to_excel('pedidos.xlsx', index=False)
```

# itens_pedido

```python
def item_pedido():
    pedido_id = random.randint(1, 1000)

    produto_id = random.randint(1, 20)

    quantidade = random.randint(1, 6)

    item_pedido = {
        "pedido_id": pedido_id,
        "produto_id": produto_id,
        "quantidade": quantidade
    }

    return item_pedido
```

```python
# Criando 2000 pedidos
lista_itens = [item_pedido() for _ in range(2000)]
df_itens_pedido = pd.DataFrame(lista_itens)
```

```python
# Adicionando a coluna de id
df_itens_pedido.insert(0, "id", range(1, len(df_itens_pedido) + 1))
```

```python
# Exportando os itens_pedidos
df_itens_pedido.to_excel('itens_pedidos.xlsx', index=False)
```

# Pagamento

```python
def pagamento():
    METODOS = ["pix", "cartao_credito", "cartao_debito", "boleto"]
    PESOS_METODOS = [0.40, 0.35, 0.15, 0.10]  # Pix e Crédito são os mais usados
    
    STATUS_PAG = ["aprovado", "pendente", "recusado"]
    PESOS_STATUS = [0.85, 0.10, 0.05]  # Amaioria é aprovada

    
    pedido_id = random.randint(1, 1000)

    metodo = random.choices(METODOS, weights=PESOS_METODOS, k=1)[0]
    status = random.choices(STATUS_PAG, weights=PESOS_STATUS, k=1)[0]

    # Gera a data do pagamento nos últimos 3 anos
    data_pagamento = fake.date_between(start_date="-3y", end_date="today").strftime("%d/%m/%Y")

    pagamento = {
        "pedido_id": pedido_id,
        "metodo": metodo,
        "status": status,
        "data_pagamento": data_pagamento,
    }

    return pagamento
```

```python
# gerando 1000 registros de pagamento
lista_pagamentos = [pagamento() for _ in range(1000)]
df_pagamentos = pd.DataFrame(lista_pagamentos)
```

```python
# Gerando a coluna de id
df_pagamentos.insert(0, "id", range(1, len(df_pagamentos) + 1))
```

```python
# Exportando essa relação
df_pagamentos.to_excel('pagamentos.xlsx', index=False)
```
